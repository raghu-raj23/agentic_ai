## Messages

Messages -> fundamental unit of context for models in LangChain
Messages -> obj with
- Role - type of message
- COntent - Message content
- Metadata - optional fields like res info, message ID, token usage, etc

LangChain provides std message type

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(model="llama3.2", model_provider="ollama")
model
```

### Text Prompts

strightforward generation no convo history

```python
model.invoke("what is AI?")
```

### Message Prompts

Can pass list of messages

Types:
- System message - Tell model how to behave and provide context for interactions
- Human message - user input and interactions
- AI message - res generated
- Tool message - output of tool call



```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on blue eyes")
]

res = model.invoke(messages)
res.content
```
