### Streaming and Batch

#### Streaming

Most LLMs can stream their output. stream() returns an iterator -> output chunks

```python
for chunk in model.stream("Write me a 200 words paragraph on Artificaial Intelligence"):
    print(chunk.text, end="|", flush=True)
```

### Batch

Collection of independent requests to a model.
Increase perfoemance and reduce cost as processing can be parallel

```python
responses = model.batch(
    [
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={
        'max_concurrency': 5,
    }
)

for response in responses:
    print(response.content)
```
